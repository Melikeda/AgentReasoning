```python
from ollama import chat
```

```python
def get_weather(city: str) -> str:
    """
    Verilen şehir için örnek hava durumu bilgisini döndürür.

    Args:
        city: Hava durumu öğrenilecek şehir.
    """
    return f"{city}: 12 derece ve parçalı bulutlu."
```

```python
messages = [
    {
        "role": "user",
        "content": "İstanbul'da bugün hava nasıl?"
    }
]
```

```python
response = chat(
    model="qwen3.5:9b",
    messages=messages,
    tools=[get_weather],
    think=True,
    stream=False
)
```

```python
messages.append(response.message)
```

```python
if response.message.tool_calls:
    tool_call = response.message.tool_calls[0]
```

```python
print("ACTION:")
print("Tool:", tool_call.function.name)
print("Argümanlar:", tool_call.function.arguments)
```

```python
result = get_weather(
    **tool_call.function.arguments
)
```

```python
print("\nOBSERVATION:")
print(result)
```

```python
messages.append(
    {
        "role": "tool",
        "tool_name": tool_call.function.name,
        "content": str(result)
    }
)
```

```python
final_response = chat(
    model="qwen3.5:9b",
    messages=messages,
    tools=[get_weather],
    think=True,
    stream=False
)
```

```python
print("\nFINAL CEVAP:")
print(final_response.message.content)
```

```python
print("\nMESAJA GEÇMİŞİ:")

for message in messages:
    if isinstance(message, dict):
        print(message["role"], "->", message.get("content", ""))
    else:
        print(message.role, "->", message.content or "Tool çağrısı")
```
