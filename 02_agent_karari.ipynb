```python
from ollama import chat
```

```python
system_prompt = """
Sen bir agent'ın karar verme bölümüsün.

Kullanıcının isteğini değerlendir.
Yalnızca bir sonraki doğru adımı belirle.

Şu üç seçenekten birini kullan:

ASK_USER: Eksik bilgiyi kullanıcıdan iste.
USE_TOOL: Kullanılması gereken tool'u yaz.
FINAL_ANSWER: Kullanıcıya doğrudan cevap ver.

Kullanabileceğin tool'lar:

get_exchange_rate:
Güncel döviz kurunu getirir.

search_flight:
Kalkış, varış ve tarih bilgilerine göre uçuş arar.
"""
```

```python
user_message = "Bana en ucuz uçuşu bul."
```

```python
response = chat(
    model="qwen3.5:9b",
    messages=[
        {
            "role": "system",
            "content": system_prompt
        },
        {
            "role": "user",
            "content": user_message
        }
    ],
    think=True,
    stream=False
)
```

```python
print(response.message.content)
```
