```python
from ollama import chat
```

```python
user_message = """
Kullanıcı şöyle dedi:

"Bana en ucuz uçuşu bul."

Bir agent olarak yalnızca bir sonraki doğru adımı belirle.
Kısa bir cevap ver.
"""
```

```python
response = chat(
    model="qwen3.5:9b",
    messages=[
        {
            "role": "user",
            "content": user_message
        }
    ],
    think=True,
    stream=False
)
```

```python
print("MODELİN CEVABI:")
print(response.message.content)
```

```python
thinking = getattr(response.message, "thinking", "")

print("REASONING:")
print(thinking)

print("\nFINAL CEVAP:")
print(response.message.content)
```
